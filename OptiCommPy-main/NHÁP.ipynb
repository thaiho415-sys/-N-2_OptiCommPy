```python
import matplotlib.patches as patches
import matplotlib.pyplot as plt

# 1. Giảm chiều cao figsize từ 3 xuống 2
fig, ax = plt.subplots(figsize=(15, 2))
ax.set_axis_off()

blocks = [
        {"text": "Dữ liệu bit", "bg": "#E3F2FD", "border": "#1E88E5"},
    {"text": "Mã hóa\n Nâng mẫu\n Định hình xung", "bg": "#E8F5E9", "border": "#43A047"},
    #{"text": "Nâng mẫu", "bg": "#E8F5E9", "border": "#43A047"},
    #{"text": "Định hình xung", "bg": "#E8F5E9", "border": "#43A047"},
    {"text": "Tín hiệu điện\n(NRZ)", "bg": "#FFF8E1", "border": "#FFB300"},
    {"text": "Bộ điều chế\nMZM", "bg": "#F3E5F5", "border": "#8E24AA"},
    {"text": "Tín hiệu quang\nOOK", "bg": "#FFEBEE", "border": "#E53935"}
]

x_offset = 0.5
width = 2.0
height = 1.0
gap = 0.8

for i, block in enumerate(blocks):
    x = x_offset + i * (width + gap)
    y = 0.8  # Đặt vị trí khối hợp lý

    rect = patches.FancyBboxPatch(
        (x, y),
        width,
        height,
        boxstyle="round,pad=0.1",
        facecolor=block["bg"],
        edgecolor=block["border"],
        linewidth=2,
    )
    ax.add_patch(rect)

    ax.text(
        x + width / 2,
        y + height / 2,
        block["text"],
        ha="center",
        va="center",
        fontsize=10,
        fontweight="bold",
    )

    if i < len(blocks) - 1:
        ax.annotate(
            "",
            xy=(x + width + gap, y + height / 2),
            xytext=(x + width, y + height / 2),
            arrowprops=dict(
                arrowstyle="-|>", lw=2, color="#424242", mutation_scale=15
            ),
        )

ax.set_xlim(0, x_offset + len(blocks) * (width + gap))

# 2. Siết chặt trục Y để bỏ khoảng trắng thừa ở trên
ax.set_ylim(0.5, 2.1)

# 3. Chỉnh pad nhỏ lại để tiêu đề nằm sát các khối
plt.title(
    "SƠ ĐỒ KHỐI PHÁT HỆ THỐNG OOK", fontsize=13, fontweight="bold", pad=2
)

plt.tight_layout()
plt.show()
```
